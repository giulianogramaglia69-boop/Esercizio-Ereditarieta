# Ciao, sono Giuliano

Esempio di classi ed ereditarietà in Python.

```python
class Animali:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta

    def __str__(self):
        return f"Ciao, mi chiamo {self.nome} e ho {self.eta} anni"
```

```python
class Cani(Animali):
    def __init__(self, nome, eta, razza):
        super().__init__(nome, eta)
        self.razza = razza

    def abbaia(self):
        print("bau")
```

```python
d1 = Cani("Andrea", 14, "pastore del caucaso")
print(d1)
d1.abbaia()
```
